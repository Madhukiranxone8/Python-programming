**DICTIONARIES**


```python
#CREATING DICTIONARIES
d={"no":18,"name":"viratkohli","runs":9300}
print(d)
print(d["name"]) #accessing a value
d["team"]="RCB" #add a new pair
print(d)
d["runs"]=9500 #modifying a value
print(d)
```

```python
#BUILT IN FUNCTIONS
dic={"name":"abhi","phNo":4873659332,"age":18,"behaviour":"hadavidi"}
print(len(dic))
print(max(dic))
print(min(dic))
print(sorted(dic))
```

```python
#write a program to check if a given key exists in a dictionary or not
k=[i for i in input().split()]
v=[i for i in input().split()]
d={i:j for i,j in zip(k,v)}
key=input()
if key in d:
  print("key is present")
else:
  print("key is not present")
```

```python
#write a program to add new key value pair to an existing  dictionary
k=[i for i in input().split()]
v=[i for i in input().split()]
d={i:j for i,j in zip(k,v)}
print(d)
d["marks"]=99
print(d)
d.update({"college":"vitb","cricket":"yes"})
print(d)
```

```python
#progarm to print sum of values in a dictionary
k=[i for i in input().split()]
v=[int(i) for i in input().split()]
d={i:j for i,j in zip(k,v)}
print(sum(d.values()))
```

```python
#write a program that prints frequency or occuarces in a list in the following
#format [1,2,3,4,3,2,1,4,5,4,3]    {1:2,2:2,3:3,4:3,5:1}
l=[int(i) for i in input().split()]
d={}
for i in l:
  d[i]=l.count(i)
print(d)
```

```python
#write a program that prints frequency or occuarces in a list in the following
#format [1,2,3,4,3,2,1,4,5,4,3]    {1:2,2:2,3:3,4:3,5:1}
#another method
l=[int(i) for i in input().split()]
d={}
for i in l:
  if i in d:
    d[i]=d[i]+1
  else:
    d[i]=1
print(d)
```
