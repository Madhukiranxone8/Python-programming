```python
#program to find largest of three number
a=int(input("enter first number:"))
b=int(input("enter second number:"))
c=int(input("enter third number:"))
if(a>b and a>c):
  print("a is largest")
elif(b>c):
  print("b is largest")
else:
  print("c is largest")
```

```python
#program to find given year is leap or not
year=int(input("enter a year"))
if(year%400==0 or (year%100!=0 and year%4==0)):
  print(f"{year} is a leap year")
else:
  print(f"{year} is not a leap year")
```

```python
#example program on MATCH CASE
day=int(input())
match day:
  case 1:
    print("monday")
  case 2:
    print("tuesday")
  case 3:
    print("wednesday")
  case 4:
    print("thursday")
  case 5:
    print("friday")
  case 6:
    print("saturday")
  case 7:
    print("sunday")
  case _:
    print("invalid")
```

```python
#WHILE LOOP
#printing 1 to 10 numbers
i=1
while(i<11):
  print(i,end=" ")
  i+=1
```

```python
#printing n to 1 numbers
n=int(input("enter a number :"))
while(n>=1):
  print(n,end=" ")
  n-=1
```

```python
#printing 1 to n even numbers
n=int(input("enter a number :"))
i=1
while(i<=n):
  if(i%2==0):
    print(i,end=" ")
  i=i+1
```

```python
#program that finds sum of digits of a given number
n=int(input())
sum=0
rem=0
while(n!=0):
  rem=n%10
  sum=sum+rem
  n=n//10
print(sum)
```

```python
#program that finds a given number is prime or not
n=int(input())
if(n<2):
  print(f"{n} is not a prime number")
else:
    for i in range(2,n):
       if(n%i==0):
         print(f"{n} is not a prime")
         break
    else:
      print(f"{n} is a prime")
```

```python
#program that prints prime numbers between given two numbers
a=int(input())
b=int(input())
for i in range(a,b+1):
  for j in range(2,i):
       if(i%j==0):
         break
  else:
      print(i,end=" ")
```
