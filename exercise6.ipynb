**TUPLES**

```python
#tuple operations
a=(1,2,3,'madhu')
b=(4,5,6,'bhagi')
print(a+b)
print(a*2)
for i in a:
  print(i)
print( 4 in a)
print(3 in a)
```

```python
#tuple methods
t=(1,2,3,3,3,'virat')
print(t.count(3))
print(t.index('virat'))
```

```python
#zip() function
t1=("n1","n2","n3")
t2=(1,2,3)
print(tuple(zip(t1,t2)))
```
