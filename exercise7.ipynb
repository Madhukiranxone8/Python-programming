**SETS IN PYTHON**

```python
#SET OPERATIONS
a={1,2,3,4}
b={4,5,6,7}
print(a|b)
print(a&b)
print(a-b)
print(a^b)
```

```python
#SET METHODS
a={1,2,3,4,5}
b={6,7,8,9,10}
a.add(6)
print(a)
a.update((7,8))
print(a)
a.remove(8)
print(a)
a.discard(7)
print(a)
a.pop()
print(a)
a.clear()
print(a)
a={1,2,3,4,5}
print(a.union(b))
print(a.intersection(b))
print(a.difference(b))
print(a.symmetric_difference(b))



```

```python
#write a python program to find the how many numbers having dupilicates in a list
l=[int(i) for i in input().split()]
l2=[]
for i in l:
  if l.count(i)>1:
     l2.append(i)
print(l2)
s=set(l2)
print(s)
print(len(s))
```
