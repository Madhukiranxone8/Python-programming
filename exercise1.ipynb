arthematic operators


```python
a=int(input("enter a number:"))
b=int(input("enter a another number:"))
print(a+b)
print(a-b)
print(a*b)
print(a/b)
print(a%b)
print(a**b)
print(a//b)
```

conditional operators

```python
c=int(input("enter a number :"))
d=int(input("enter a another number :"))
print(c==d)
print(c!=d)
print(c>d)
print(c<d)
print(c<=d)
print(c>=d)

```

logical operators

```python
x=int(input())
y=int(input())
print(x>y and x<y)
print(x<=y or x>=y)
print(not y)
```
