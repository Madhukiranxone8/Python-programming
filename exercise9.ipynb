```python
#copying content of one file to another
with open("madhu.txt","r") as f:
  data=f.read()
  print("reading done")
with open("madhu2.txt","w") as f2:
  f2.write(data)
  print("file copied")
```

```python
#program that prints each line of a file in reverse order
with open("madhu.txt","r") as f:
  data=f.readlines()
  for i in data:
    print(i[::-1])
```

```python
#write a program to sort words in a file and put them in another file.the output
#file should have only lower case words, so any upper case from source must be lowered
with open("madhu.txt","r") as f:
  data==f.read().split()
  data_lower=[i .lower() for i in data]
  sort_data=sorted(data_lower)
with open("madhu3.txt","w") as f2:
  for i in sort_data:
    f2.write(i+"\n")
  print("done")
```

```python
# python program to compute the number of characters ,words and lines in a file
file =open("madhu.txt","r")
characters=0
words=0
lines=0
for line in file:
  lines  += 1
  characters += len(line)
  words += len(line.split())
file.close()
print("number of characters:",characters)
print("number of words:",words)
print("number of lines:",lines)
```

```python
#rb,wb,ab
with open("pic.jpg","rb") as f:
  data=f.read()
  print(data)
with open("pic2.jpg","wb") as f2:
  f2.write(data)
  print("pic created")
```

```python
import pickle
student={"name":"madhu","age":18}
with open("student.dat","wb") as f:
  pickle.dump(student,f)
print("data written successfully.")
with open("student.dat","rb") as f2:
  data=pickle.load(f2)
print("student data :")
print(data)
```
